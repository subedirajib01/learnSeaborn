```python
sns.set_style("ticks")
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island")
sns.despine(left=True)
```

```python
sns.set_style("ticks")
sns.set_context("paper")
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island")
sns.despine()
```

The set_context() can be paper, poster, talk, and notebook

```python
sns.set_style("ticks")
sns.set_context("paper")
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island",palette="Dark2")
sns.despine()
# Here set1 gives brighter color
# set2 gives not that much brighter color
# set3 gives faded color
# Others can be Dark2,
```

```python
sns.scatterplot(data=penguins,x="species",y="body_mass_g",hue="island",style="sex",alpha=0.5)
# Here alpha will descrease the opacity and whether there is overlap the opacity will be increase

# The style parameter in sns.scatterplot is used to map different marker styles
#  (like different shapes) to the data points based on a categorical variable.
# In your code, style="sex" means that the points for male and female penguins
# will be represented with different marker shapes, making it easier to distinguish
#  between the two sexes on the plot.
```

```python
sns.stripplot(data=penguins,x="species",y="body_mass_g",hue="island")
```

```python
sns.stripplot(data=penguins,x="species",y="body_mass_g",hue="island",dodge=True)
# The dodge=True parameter in sns.stripplot separates or offsets the points
#  for different categories within the same x-axis position,
#  preventing overlap and making it easier to see the distribution of each category.
```

```python
sns.stripplot(data=penguins,x="species",y="body_mass_g",hue="island",dodge=True,jitter=True)
# The jitter=True parameter in sns.stripplot adds a small amount of random noise to
#  the x-axis positions of the points. This helps to spread out overlapping points,
  # making it easier to see the density of data within each category,
#  especially when many points fall at the same x-coordinate.
```

```python
sns.set_context("talk")
sns.swarmplot(data=penguins,x="species",y="body_mass_g",hue="island",dodge=True,palette="Set1")
sns.despine()
# The sns.despine() function is used to remove the top and right spines
#  (the lines that form the borders of the plot)
```

```python
# histogram
sns.set_context("notebook")
sns.histplot(data=penguins,x="body_mass_g",hue="sex",multiple="stack")
# The hue='sex' parameter in sns.histplot separates the data into different distributions
#  based on the unique values in the 'sex' column (Male and Female),
#  and multiple='stack' then stacks these distributions on top of each other within each bin.
# This allows you to see the contribution
#  of each 'sex' to the total count for a given body mass range.
```

```python
# regeression plot
sns.regplot(data=penguins,x="body_mass_g",y="flipper_length_mm",color="green",scatter=True)
# Scatter=False will make the data points disable
```

```python
# lineplot
sns.set_style("whitegrid")
sns.lineplot(data=penguins,x="body_mass_g",y="flipper_length_mm",hue="island",style="sex")
```

```python
# Jointplot:help to combine two different plot
sns.set_context("notebook")
sns.set_style("ticks")
sns.jointplot(data=penguins,x="body_mass_g",y="flipper_length_mm",kind="reg")
sns.despine()
# kind` must be one of ['scatter', 'hist', 'hex', 'kde', 'reg', 'resid']
```

```python
sns.set_context("notebook")
sns.set_style("whitegrid")
sns.barplot(data=penguins,x="species",y="body_mass_g",hue="sex",palette=["pink","skyblue"],estimator=np.median)
# estimator=np.sum can be np.median,np.std,np.var
```

```python
# countplot
sns.set_context("notebook")
sns.countplot(data=penguins,x="species",hue="sex",palette="Set2")
```

```python
# box-plot
sns.boxplot(data=penguins,x="species",y="body_mass_g",hue="sex",palette=["green","blue"])
```

```python
# Violin plot
sns.violinplot(data=penguins,x="species",y="body_mass_g",hue="sex",palette="Set1",split=True,inner="box")
# split help here to split the data between male and female
```

```python
sns.violinplot(data=penguins,x="species",y="body_mass_g",palette="pastel")
sns.swarmplot(data=penguins,x="species",y="body_mass_g",color="black",size=3,hue="island",palette='dark')
```

```python
# KDE -> for estimating the probability density function of continuous or non-parametric data
# Or in simple terms: a smooth curve that shows the distribution of data- like a smoothed-out version
# It helps you see where the data is concentrated (peaks) and how it spreads.

sns.kdeplot(data=penguins,x="body_mass_g",hue="sex",fill=True,alpha=0.5)
sns.despine()
```

```python
# Heatmap: a 2D color coded matrix that shows relationships or patterns between two variables.
# It's commonly used to:
# - show correlation between features
# - visualize confusion matrices- will see in the later stages
# - Highlight missing data -> sns.heatmap(df.isnull())

columns=["bill_length_mm","bill_depth_mm",	"flipper_length_mm",	"body_mass_g"]
penguins[columns].corr()
```

```python
columns=["bill_length_mm","bill_depth_mm",	"flipper_length_mm",	"body_mass_g"]
sns.heatmap(penguins[columns].corr(),annot=True,cmap="Blues",vmin=-0.2,linewidth=2,linecolor="black")
```

```python
# Rug-plot:
sns.rugplot(data=penguins,y="body_mass_g",hue="species",palette="Set2",height=0.2)
```

```python
# Pair-plot:
sns.pairplot(data=penguins,hue="sex",palette="Set2",diag_kind="hist",)
# here in diag-kind, it can be kde, hist,
```

```python
# Pair grid
# subplot grid for plotting pair wise relationships in a dataset

graph=sns.PairGrid(data=penguins,hue="species",palette="Set2")
graph.map_upper(sns.scatterplot) #upper triangle
graph.map_lower(sns.kdeplot) #lower triangle
graph.map_diag(sns.histplot) # diagonal
graph.add_legend()
```
